## Task 1

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import f1_score
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
import pickle

# Step 1: Encode string labels to numeric labels (fake = 0, reliable = 1)
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)  # Encode training labels
y_val_encoded = label_encoder.transform(y_val)          # Encode validation labels

# Step 2: Custom tokenizer function for tokenized content
def custom_tokenizer(doc):
    return doc  # Here `doc` is already tokenized, so return it as is

# Initialize CountVectorizer with custom tokenizer
vectorizer = CountVectorizer(tokenizer=custom_tokenizer, lowercase=False, max_features=10000)

print(X_train)

# Vectorize the tokenized training and validation data
X_train_vec = vectorizer.fit_transform(X_train)  # Vectorizing training data
X_val_vec = vectorizer.transform(X_val)  # Vectorizing validation data



log_reg = LogisticRegression(max_iter=10000)

# Fit the model to the vectorized training data
log_reg.fit(X_train_vec, y_train_encoded)

# Save the trained model
with open('logreg1.pkl', 'wb') as f:
    pickle.dump(log_reg, f)

# Step 4: Evaluate the model on the validation data
y_pred = log_reg.predict(X_val_vec)

# Calculate F1 score using the encoded labels
f1 = f1_score(y_val_encoded, y_pred)

# Output the results
print(f"F1 Score: {f1:.4f}")
```

```python
from sklearn.metrics import classification_report

label_values = ["fake", "reliable"]
print(classification_report(y_val_encoded, y_pred, target_names=[str(l) for l in label_values]))
```

## Task 2

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import f1_score
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import CountVectorizer
from scipy.sparse import hstack
import pickle

# Ensure tokenized_content is properly formatted
def fix_tokenized_content(df):
    df['tokenized_content'] = df['tokenized_content'].apply(lambda x: ' '.join(x) if isinstance(x, list) else '')
    return df

# Apply fix to both train and validation sets
X_train_combined = fix_tokenized_content(X_train.to_frame())
X_val_combined = fix_tokenized_content(X_val.to_frame())

# Add domain
X_train_combined['domain'] = large_set.loc[X_train_combined.index, 'domain']
X_val_combined['domain'] = large_set.loc[X_val_combined.index, 'domain']

print(X_train_combined['tokenized_content'])

# Vectorize tokenized content (use sparse representation)
vectorizer = CountVectorizer(max_features=10000)
X_train_text_features = vectorizer.fit_transform(X_train_combined['tokenized_content'])
X_val_text_features = vectorizer.transform(X_val_combined['tokenized_content'])



# Calculate URL and Date counts
X_train_meta = np.vstack([
    X_train_combined['tokenized_content'].apply(fn.count_urls),
    X_train_combined['tokenized_content'].apply(fn.count_dates)
]).T

X_val_meta = np.vstack([
    X_val_combined['tokenized_content'].apply(fn.count_urls),
    X_val_combined['tokenized_content'].apply(fn.count_dates)
]).T

# One-hot encode domain (keep sparse)
encoder = OneHotEncoder(handle_unknown='ignore')
X_train_domain = encoder.fit_transform(X_train_combined[['domain']])
X_val_domain = encoder.transform(X_val_combined[['domain']])

# Combine all features (sparse representation)
X_train_final = hstack([X_train_text_features, X_train_meta, X_train_domain])
X_val_final = hstack([X_val_text_features, X_val_meta, X_val_domain])

# Train sparse-aware Logistic Regression
model_meta = LogisticRegression(max_iter=1000)
model_meta.fit(X_train_final, y_train_encoded)

# Save the trained model
with open('logreg1.pkl', 'wb') as f:
    pickle.dump(model_meta, f)

# Predict and evaluate
y_pred_meta = model_meta.predict(X_val_final)

# F1 Score
f1_meta = f1_score(y_val_encoded, y_pred_meta)
print(f"F1 Score: {f1_meta:.4f}")
```

```python
from sklearn.metrics import classification_report

label_values = ["fake", "reliable"]
print(classification_report(y_val_encoded, y_pred_meta, target_names=[str(l) for l in label_values]))
```
