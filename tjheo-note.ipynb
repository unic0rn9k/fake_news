## Task 1

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import f1_score
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
import pickle

# Step 1: Encode string labels to numeric labels (fake = 0, reliable = 1)
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)  # Encode training labels
y_val_encoded = label_encoder.transform(y_val)          # Encode validation labels

# Step 2: Custom tokenizer function for tokenized content
def custom_tokenizer(doc):
    return doc  # Here `doc` is already tokenized, so return it as is

# Initialize CountVectorizer with custom tokenizer
vectorizer = CountVectorizer(tokenizer=custom_tokenizer, lowercase=False, max_features=10000)

# Vectorize the tokenized training and validation data
X_train_vec = vectorizer.fit_transform(X_train)  # Vectorizing training data
X_val_vec = vectorizer.transform(X_val)  # Vectorizing validation data

# Step 3: Logistic Regression without Hyperparameter Tuning
log_reg = LogisticRegression(max_iter=10000)

# Fit the model to the vectorized training data
log_reg.fit(X_train_vec, y_train_encoded)

# Save the trained model
with open('logreg1.pkl', 'wb') as f:
    pickle.dump(log_reg, f)

# Step 4: Evaluate the model on the validation data
y_pred = log_reg.predict(X_val_vec)

# Calculate F1 score using the encoded labels
f1 = f1_score(y_val_encoded, y_pred)

# Output the results
print(f"F1 Score: {f1:.4f}")
```

```python
from sklearn.metrics import classification_report

label_values = ["fake", "reliable"]
print(classification_report(y_val_encoded, y_pred, target_names=[str(l) for l in label_values]))
```

## Task 2

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.metrics import f1_score
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from scipy.sparse import hstack
import pickle

# Ensure 'tokenized_content' is a list of words
def fix_tokenized_content(df):
    df['tokenized_content'] = df['tokenized_content'].apply(lambda x: x if isinstance(x, list) else [])
    return df

def extract_manual_features(df):
    df = fix_tokenized_content(df)  # Fix any non-list values in 'tokenized_content'
    df['char_count'] = df['tokenized_content'].str.len()
    df['word_count'] = df['tokenized_content'].apply(len)  # Apply len directly to list of tokens
    df['avg_word_len'] = df['char_count'] / (df['word_count'] + 1)
    df['exclamation_count'] = df['tokenized_content'].str.count('!')
    df['capital_count'] = df['tokenized_content'].apply(lambda x: sum(1 for word in x if word.isupper()))
    return df[['char_count', 'word_count', 'avg_word_len', 'exclamation_count', 'capital_count']].values

# Step 1: Add meta-data (domain, url_count, date_count) to X_train and X_val
X_train_combined = X_train.to_frame()  # Convert Series to DataFrame (with 'content')
X_val_combined = X_val.to_frame()      # Same for validation

# Add 'domain' (from original data, assuming it's in your test_set DataFrame)
X_train_combined['domain'] = large_set['domain']
X_val_combined['domain'] = large_set['domain']

X_train_text_features = extract_manual_features(X_train_combined)
X_val_text_features = extract_manual_features(X_val_combined)

# Step 3: Add meta-data (url_count, date_count)
X_train_combined['url_count'] = X_train_combined['tokenized_content'].apply(fn.count_urls)
X_train_combined['date_count'] = X_train_combined['tokenized_content'].apply(fn.count_dates)

X_val_combined['url_count'] = X_val_combined['tokenized_content'].apply(fn.count_urls)
X_val_combined['date_count'] = X_val_combined['tokenized_content'].apply(fn.count_dates)

X_train_meta = X_train_combined[['url_count', 'date_count']].values
X_val_meta = X_val_combined[['url_count', 'date_count']].values

# Step 4: One-hot encode domain
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

X_train_domain = encoder.fit_transform(X_train_combined[['domain']])
X_val_domain = encoder.transform(X_val_combined[['domain']])

# Combine features
X_train_final = np.hstack([X_train_text_features, X_train_meta, X_train_domain])
X_val_final = np.hstack([X_val_text_features, X_val_meta, X_val_domain])

# Step 5: Handle missing values (NaN)
# Use SimpleImputer to fill missing values with the mean for numeric features
imputer = SimpleImputer(strategy='mean')

X_train_final_imputed = imputer.fit_transform(X_train_final)
X_val_final_imputed = imputer.transform(X_val_final)

# Step 6: Train and evaluate
model_meta = LogisticRegression(max_iter=1000)
model_meta.fit(X_train_final_imputed, y_train_encoded)

with open('logreg1.pkl', 'wb') as f:
    pickle.dump(model_meta, f)  # Save the trained model
    
y_pred_meta = model_meta.predict(X_val_final_imputed)

f1_meta = f1_score(y_val_encoded, y_pred_meta)
print(f"F1 Score: {f1_meta:.4f}")
```

```python
from sklearn.metrics import classification_report

label_values = ["fake", "reliable"]
print(classification_report(y_val_encoded, y_pred_meta, target_names=[str(l) for l in label_values]))
```
