## Task 1

```python
# Define types that are considered REAL (label 0)
real_types = ['reliable', 'political']

# Create label column
test_set['label'] = test_set['type'].apply(lambda x: 0 if str(x).lower() in real_types else 1)
```

```python
# Step 1: Create a copy of test_set so you don't overwrite your real 'type' column
token_df = test_set.copy()

# Step 2: Overwrite 'type' column with dummy label
token_df['type'] = 'all'

# Step 3: Get top 10,000 words across entire dataset
raw_vocab = fn.word_freq(token_df, top_k=10000)
vocab_list = list(raw_vocab["all"].keys())
vocab_dict = {word: i for i, word in enumerate(vocab_list)}
```

```python
from sklearn.feature_extraction.text import CountVectorizer


vectorizer = CountVectorizer(vocabulary=vocab_dict)
X = vectorizer.fit_transform(token_df['content'])  # Use original content
y = token_df['label']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=30)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.model_selection import GridSearchCV


#Logistic Regression model with hyperparameter tuning options
log_reg = LogisticRegression(max_iter=10000)

param_grid = {
    'C': [0.001, 0.01, 0.1, 1, 10],          
    'solver': ['liblinear', 'saga', 'newton-cg'], 
    'penalty': ['l2'],                        
}

# Performs GridSearchCV to find the best hyperparameters
grid_search = GridSearchCV(log_reg, param_grid, cv=5, scoring='f1', n_jobs=-1)

# Fits the model to the training data
grid_search.fit(X_train, y_train)

# Saves the best model from the grid search
best_model = grid_search.best_estimator_

# Evaluate the best model on the test data
y_pred = best_model.predict(X_test)
f1 = f1_score(y_test, y_pred)

print(f"Best Hyperparameters: {grid_search.best_params_}")
print(f"F1 Score: {f1:.4f}")
```

## Task 2

```python
# Step 1: Split early
train_set_meta, val_set_meta = train_test_split(test_set, test_size=0.2)

# Step 2: Build vocab only on train
raw_vocab_meta = fn.word_freq(token_df, top_k=10000)
vocab_list_meta = list(raw_vocab_meta["all"].keys())
vocab_dict_meta = {word: i for i, word in enumerate(vocab_list_meta)}

# Step 3: Fit TF-IDF vectorizer on train content only
vectorizer = TfidfVectorizer(vocabulary=vocab_dict_meta)
X_train_text_meta = vectorizer.fit_transform(train_set_meta['content'])
X_test_text_meta = vectorizer.transform(val_set_meta['content'])


from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=True) 
X_train_domain_meta = encoder.fit_transform(train_set_meta[['domain']])
X_test_domain_meta = encoder.transform(val_set_meta[['domain']])

# Step 5: Meta features
train_set_meta['url_count'] = train_set_meta['content'].apply(fn.count_urls)
train_set_meta['date_count'] = train_set_meta['content'].apply(fn.count_dates)
val_set_meta['url_count'] = val_set_meta['content'].apply(fn.count_urls)
val_set_meta['date_count'] = val_set_meta['content'].apply(fn.count_dates)

X_train_meta = train_set_meta[['url_count','date_count']].values
X_test_meta = val_set_meta[['url_count','date_count']].values

# Step 6: Combine everything
from scipy.sparse import hstack

X_train_meta = hstack([X_train_text_meta, X_train_meta, X_train_domain_meta])
X_test_meta = hstack([X_test_text_meta, X_test_meta, X_test_domain_meta])

y_train_meta = train_set_meta['label']
y_test_meta = val_set_meta['label']

# Step 7: Train and evaluate
model_meta = LogisticRegression(max_iter=1000)
model_meta.fit(X_train_meta, y_train_meta)
y_pred_meta = model_meta.predict(X_test_meta)

from sklearn.metrics import f1_score
f1_meta = f1_score(y_test_meta, y_pred_meta)
print(f"F1 Score: {f1_meta:.4f}")
```
