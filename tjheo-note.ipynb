## Task 1

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import f1_score
from collections import Counter
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV

# Step 1: Encode string labels to numeric labels (fake = 0, reliable = 1)
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)  # Encode training labels
y_val_encoded = label_encoder.transform(y_val)          # Encode validation labels

# Step 2: Create vocabulary (top 10,000 most frequent words across entire dataset)
raw_vocab = fn.word_freq(large_set, top_k=10000)  # Use test_set directly
combined_vocab = Counter()

# Combine all word frequencies from all types
for type_, words in raw_vocab.items():
    combined_vocab.update(words)

# Get the top 10,000 most frequent words
top_10000_words = [word for word, _ in combined_vocab.most_common(10000)]

# Create the vocabulary dictionary
vocab_dict = {word: i for i, word in enumerate(top_10000_words)}

# Step 3: Vectorize text data using the created vocabulary
vectorizer = CountVectorizer(vocabulary=vocab_dict)
X_train_vec = vectorizer.fit_transform(X_train)
X_val_vec = vectorizer.transform(X_val)

# Step 4: Logistic Regression with Hyperparameter Tuning
log_reg = LogisticRegression(max_iter=10000)


param_grid = {
    'C': [0.001, 0.01, 0.1, 1, 10],
    'solver': ['liblinear', 'saga', 'newton-cg'],
    'penalty': ['l2'],
}

# Perform GridSearchCV to find the best hyperparameters
grid_search = GridSearchCV(log_reg, param_grid, cv=5, scoring='f1', n_jobs=-1)

# Fit the model to the training data
grid_search.fit(X_train_vec, y_train_encoded)

with open('logreg1.pkl', 'wb') as f:
    pickle.dump(log_reg, f)
    
# Save the best model from the grid search
best_model = grid_search.best_estimator_

# Step 5: Evaluate the best model on the validation data
y_pred = best_model.predict(X_val_vec)

# Calculate F1 score using the encoded labels
f1 = f1_score(y_val_encoded, y_pred)

# Output the results
print(f"Best Hyperparameters: {grid_search.best_params_}")
print(f"F1 Score: {f1:.4f}")
```

```python
from sklearn.preprocessing import OneHotEncoder
from scipy.sparse import hstack
from sklearn.metrics import f1_score

def extract_manual_features(df):
    df['char_count'] = df['content'].str.len()
    df['word_count'] = df['content'].str.split().apply(len)
    df['avg_word_len'] = df['char_count'] / (df['word_count'] + 1)
    df['exclamation_count'] = df['content'].str.count('!')
    df['capital_count'] = df['content'].str.findall(r'[A-Z]').apply(len)
    return df[['char_count', 'word_count', 'avg_word_len', 'exclamation_count', 'capital_count']].values

# Step 1: Add meta-data (domain, url_count, date_count) to X_train and X_val
X_train_combined = X_train.to_frame()  # Convert Series to DataFrame (with 'content')
X_val_combined = X_val.to_frame()      # Same for validation

# Add 'domain' (from original data, assuming it's in your test_set DataFrame)
X_train_combined['domain'] = large_set['domain']
X_val_combined['domain'] = large_set['domain']

X_train_text_features = extract_manual_features(X_train_combined)
X_val_text_features = extract_manual_features(X_val_combined)

# Step 3: Add meta-data (url_count, date_count)
X_train_combined['url_count'] = X_train_combined['content'].apply(fn.count_urls)
X_train_combined['date_count'] = X_train_combined['content'].apply(fn.count_dates)

X_val_combined['url_count'] = X_val_combined['content'].apply(fn.count_urls)
X_val_combined['date_count'] = X_val_combined['content'].apply(fn.count_dates)

X_train_meta = X_train_combined[['url_count', 'date_count']].values
X_val_meta = X_val_combined[['url_count', 'date_count']].values

# Step 4: One-hot encode domain
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

X_train_domain = encoder.fit_transform(X_train_combined[['domain']])
X_val_domain = encoder.transform(X_val_combined[['domain']])

# Step 5: Combine features
X_train_final = np.hstack([X_train_text_features, X_train_meta, X_train_domain])
X_val_final = np.hstack([X_val_text_features, X_val_meta, X_val_domain])

# Step 6: Train and evaluate
model_meta = LogisticRegression(max_iter=1000)
model_meta.fit(X_train_final, y_train_encoded)

with open('logreg2.pkl', 'wb') as f:
    pickle.dump(model_meta, f)
    
y_pred_meta = model_meta.predict(X_val_final)

f1_meta = f1_score(y_val_encoded, y_pred_meta)
print(f"F1 Score: {f1_meta:.4f}")
```
