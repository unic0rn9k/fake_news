```python
df = pd.read_csv("cleaned_data.csv").dropna()
df
```

The following code contains a dictionary that maps the original labels of the dataset, to the desired label format.
Here the None label refers to content that should be omitted entirely from the dataset.
For grouping the original labels, I followed a very simple algorithm.
I refferred to the source referenced in the assignment (https://github.com/several27/FakeNewsCorpus)
and marked any label that explicitly was mentioned to contain exclusively incorrect / non-factual information as "fake",
while marking labels that explicitly was mentioned to generally only contain factually verifiable information as "reliable",
and lastly marking anything that was neither as None (meaning it was implied, or explicitly stated that it might contain some mix of factually verifiable, and "fake" information).

```python
label_map = {
    "fake": "fake",
    "satire": None,
    "bias": None,
    "conspiracy": "fake",
    "state": None,
    "junksci": "fake",
    "hate": None,
    "clickbait": None,
    "unreliable": None,
    "political": "reliable",
    "reliable": "reliable",
    "unknown": None,
}

df["new_labels"] = [label_map[n] for n in df["type"]]
```

```python
nfake = sum(df["new_labels"] == "fake")
nreliable = sum(df["new_labels"] == "reliable")

print(f"{nfake/(nreliable+nfake) * 100}% fake news articles")
```

In this dataset, the overwhelmingly large majority of the data is labeled "fake".
In classification tasks, an imbalanced distribution of labels can lead to biased models that perform well on the majority class but poorly on the minority class. This imbalance means that the model may not learn enough about the characteristics of the minority class, resulting in reduced predictive accuracy and potentially discriminatory outcomes. Techniques such as oversampling the minority class, undersampling the majority class, or using algorithms designed for imbalanced data can help mitigate these risks.
